```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df= pd.read_csv("AI Job Market Dataset.csv")
df.shape
```

```python
df.shape
df.columns
df.info()
df.describe()
```

```python
#Missing Values
df.isnull().sum()
```

```python
#handling Missing Values
df = df.dropna()
df.fillna(0, inplace=True)
```

```python
#remove Duplicates
df = df.drop_duplicates()
df
```

```python
#salary Analysis
print(df['salary'].mean())
print(df['salary'].max())
print(df['salary'].min())
```

```python
# Groups the Data by job role calculates average salary per role
salary_by_role = df.groupby('job_title')['salary'].mean().sort_values(ascending=False)
print(salary_by_role)
```

```python
import matplotlib.pyplot as plt
salary_by_role.plot(kind='barh')
plt.xlabel('salary by role')
plt.ylabel('Job_roles')
plt.title("Analysis Jobs roles")
plt.show()
```

```python
#how salary  increases by experience
exp_salary = df.groupby('experience_level')['salary'].mean()
print(exp_salary)
```

```python
company_salary = df.groupby('company_industry')['salary'].mean().sort_values(ascending=False)
print(company_salary)
```

```python
import matplotlib.pyplot as plt
company_salary.plot(kind='line')
plt.xlabel('salary')
plt.ylabel('company_industry')
plt.title('Company Expenditure')
plt.legend()
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt
salary_by_role.head(10).plot(kind='pie')
plt.title("Top 10 job roles by salary")
plt.show()
```
